## 2. Feature Engineering
Creating lag features (previous weeks' sales), rolling averages, and date-derived features (week of year, month, holiday flags). These become the core predictive signals for the ML models.

```python
# Total markdown (fill NaN with 0 first — missing markdown just means no promo that week)
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']
features[markdown_cols] = features[markdown_cols].fillna(0)
features['Total_MarkDown'] = features[markdown_cols].sum(axis=1)

# Store type dummies
store_type_dummies = pd.get_dummies(stores['Type'], prefix='Store_Type', drop_first=True)
stores = pd.concat([stores, store_type_dummies], axis=1)

# Merge
train = train.merge(stores, on='Store', how='left').merge(
    features.drop(columns=markdown_cols), on=['Store','Date'], how='left')
test = test.merge(stores, on='Store', how='left').merge(
    features.drop(columns=markdown_cols), on=['Store','Date'], how='left')

# IsHoliday can appear in both train and the merged features frame -> collapse duplicate columns
for df in [train, test]:
    if 'IsHoliday_y' in df.columns:
        df['IsHoliday'] = df['IsHoliday_x']
        df.drop(columns=['IsHoliday_x','IsHoliday_y'], inplace=True)
```

```python
# Date-derived features
for df in [train, test]:
    df['Month'] = df['Date'].dt.month
    df['Year'] = df['Date'].dt.year
    df['WeekOfYear'] = df['Date'].dt.isocalendar().week.astype(int)
    df['Quarter'] = df['Date'].dt.quarter

# Sort chronologically WITHIN each Store-Dept before computing lag/rolling features
train = train.sort_values(['Store','Dept','Date']).reset_index(drop=True)

train['Weekly_Sales_Lag1'] = train.groupby(['Store','Dept'])['Weekly_Sales'].shift(1)
train['Weekly_Sales_Lag2'] = train.groupby(['Store','Dept'])['Weekly_Sales'].shift(2)
train['Rolling_Mean_4'] = train.groupby(['Store','Dept'])['Weekly_Sales'].transform(
    lambda s: s.shift(1).rolling(window=4).mean())
train['Rolling_Std_4'] = train.groupby(['Store','Dept'])['Weekly_Sales'].transform(
    lambda s: s.shift(1).rolling(window=4).std())

# Fill remaining NaNs (early rows of each group with no history yet, and any leftover feature NaNs)
fill_cols = ['CPI','Unemployment','Temperature','Fuel_Price','Total_MarkDown',
             'Weekly_Sales_Lag1','Weekly_Sales_Lag2','Rolling_Mean_4','Rolling_Std_4']
train[fill_cols] = train[fill_cols].fillna(train[fill_cols].median())
test[['CPI','Unemployment','Temperature','Fuel_Price','Total_MarkDown']] = \
    test[['CPI','Unemployment','Temperature','Fuel_Price','Total_MarkDown']].fillna(
        train[['CPI','Unemployment','Temperature','Fuel_Price','Total_MarkDown']].median())

train.head()
```

## 5. Stationarity Check & Seasonal Decomposition
ARIMA/SARIMA assume (near-)stationary data. Rather than assuming, we test with the
**Augmented Dickey-Fuller test** on a representative Store-Dept series and visualize the decomposition.
This justifies *why* differencing/seasonal terms are needed instead of guessing (p,d,q).

```python
def adf_report(series, label=""):
    result = adfuller(series.dropna())
    print(f"ADF Statistic ({label}): {result[0]:.4f}")
    print(f"p-value: {result[1]:.4f}")
    if result[1] < 0.05:
        print("=> Series is likely STATIONARY (reject H0)")
    else:
        print("=> Series is likely NON-STATIONARY (fail to reject H0) — differencing needed")
    print()

sample_store, sample_dept = 1, 1
sample_series = train[(train['Store']==sample_store) & (train['Dept']==sample_dept)] \
    .set_index('Date')['Weekly_Sales'].asfreq('W-FRI').fillna(method='ffill')

adf_report(sample_series, f"Store {sample_store} Dept {sample_dept} - Level")
adf_report(sample_series.diff(1), f"Store {sample_store} Dept {sample_dept} - 1st Difference")
```

```python
decomp = seasonal_decompose(sample_series, model='additive', period=52)
fig = decomp.plot()
fig.set_size_inches(12, 8)
plt.tight_layout(); plt.show()
# Clear yearly seasonal component -> supports using a seasonal order (SARIMA) over plain ARIMA
```

## 6. Evaluation Metric (WMAE) & Train/Val/Test Split
WMAE weights holiday weeks 5x, matching Walmart's actual business cost. 3-way chronological split created here — see verification note below for split dates.

```python
def wmae(y_true, y_pred, is_holiday):
    weights = np.where(is_holiday, 5, 1)
    return np.sum(weights * np.abs(np.array(y_true) - np.array(y_pred))) / np.sum(weights)

# 3-way chronological split (all carved out of train.csv, since Kaggle's test.csv has no
# Weekly_Sales labels and can't be used for evaluation). Kept strictly time-ordered to avoid leakage:
#   tr        -> used to fit models
#   val       -> used for model selection / comparison / ablation / error analysis (as before)
#   test_eval -> held out completely until the very end; used ONLY for final, unbiased evaluation
test_cutoff = train['Date'].max() - pd.Timedelta(weeks=12)   # last 12 weeks -> held-out TEST
val_cutoff  = test_cutoff - pd.Timedelta(weeks=12)             # 12 weeks before that -> VAL

tr = train[train['Date'] <= val_cutoff].copy()
val = train[(train['Date'] > val_cutoff) & (train['Date'] <= test_cutoff)].copy()
test_eval = train[train['Date'] > test_cutoff].copy()

print(f"Train:     {tr['Date'].min().date()} to {tr['Date'].max().date()}  ({len(tr)} rows)")
print(f"Val:       {val['Date'].min().date()} to {val['Date'].max().date()}  ({len(val)} rows)")
print(f"Test_eval: {test_eval['Date'].min().date()} to {test_eval['Date'].max().date()}  ({len(test_eval)} rows)")
```

## 7. Baseline Model — Seasonal Naive
Predicts each week as the historical average for that Store-Dept-WeekOfYear.

```python
def seasonal_naive_predict(tr_df, val_df):
    lookup = tr_df.set_index(['Store','Dept','WeekOfYear'])['Weekly_Sales'].groupby(
        level=[0,1,2]).mean()
    preds = []
    for _, row in val_df.iterrows():
        key = (row['Store'], row['Dept'], row['WeekOfYear'])
        preds.append(lookup.get(key, tr_df['Weekly_Sales'].median()))
    return np.array(preds)

baseline_preds = seasonal_naive_predict(tr, val)
baseline_wmae = wmae(val['Weekly_Sales'], baseline_preds, val['IsHoliday'])
print(f"Baseline (Seasonal Naive) WMAE: {baseline_wmae:,.2f}")
```

## 8. XGBoost Model
Gradient-boosted tree model trained on engineered features.

```python
from joblib import Parallel, delayed
from statsmodels.tools.sm_exceptions import ConvergenceWarning
from tqdm.notebook import tqdm
warnings.filterwarnings("ignore", category=ConvergenceWarning)

def fit_sarima_forecast(series, steps):
    try:
        model = SARIMAX(series, order=(1,1,1), seasonal_order=(1,1,1,52),
                         enforce_stationarity=False, enforce_invertibility=False)
        fit = model.fit(disp=False, maxiter=50, method='lbfgs')
        return fit.forecast(steps=steps).values
    except Exception:
        return None

MIN_HISTORY_WEEKS = 104
TOP_N_COMBOS = 200

combo_volume = tr.groupby(['Store','Dept'])['Weekly_Sales'].sum().sort_values(ascending=False)
top_combos = set(combo_volume.head(TOP_N_COMBOS).index)

print(f"Fitting SARIMA on top {TOP_N_COMBOS} combos "
      f"(covering {combo_volume.head(TOP_N_COMBOS).sum() / combo_volume.sum():.1%} of training sales volume)")

# --- Precompute small, self-contained payloads BEFORE parallelizing ---
# Passing the full `tr` / `grouped_tr` into every worker call makes joblib repeatedly
# pickle/memmap huge objects to disk, which is what filled up the disk last time.
# Instead we build a lightweight task list: just the series (or None) and dates needed.
grouped_tr = tr.groupby(['Store','Dept'])
tasks = []

for (store, dept), v_group in val.groupby(['Store','Dept']):
    v_group = v_group.sort_values('Date')
    dates = v_group['Date'].values
    steps = len(v_group)

    series = None
    if (store, dept) in top_combos and (store, dept) in grouped_tr.groups:
        t_group = grouped_tr.get_group((store, dept)).sort_values('Date')
        s = t_group.set_index('Date')['Weekly_Sales'].asfreq('W-FRI').fillna(method='ffill')
        if len(s) >= MIN_HISTORY_WEEKS:
            series = s

    fallback = seasonal_naive_predict(tr, v_group)

    tasks.append((store, dept, dates, steps, series, fallback))

def process_task(store, dept, dates, steps, series, fallback):
    if series is not None:
        forecast = fit_sarima_forecast(series, steps)
        if forecast is not None:
            return pd.DataFrame({'Store': store, 'Dept': dept, 'Date': dates, 'SARIMA_Pred': forecast})
    return pd.DataFrame({'Store': store, 'Dept': dept, 'Date': dates, 'SARIMA_Pred': fallback})

# max_nbytes='1M' forces small arrays to be passed directly instead of memmapped to disk
results = Parallel(n_jobs=-1, backend='loky', max_nbytes='1M')(
    delayed(process_task)(*task) for task in tqdm(tasks, desc="Fitting SARIMA (scoped)")
)

sarima_preds_df = pd.concat(results)
val_sarima = val.merge(sarima_preds_df, on=['Store','Dept','Date'], how='left')
val_sarima['SARIMA_Pred'] = val_sarima['SARIMA_Pred'].fillna(val_sarima['Weekly_Sales'].median())

sarima_wmae = wmae(val_sarima['Weekly_Sales'], val_sarima['SARIMA_Pred'], val_sarima['IsHoliday'])
print(f"SARIMA (top {TOP_N_COMBOS} combos, seasonal-naive fallback elsewhere) WMAE: {sarima_wmae:,.2f}")
```

## 9. Feature Importance
  Which engineered features XGBoost relies on most.

```python
feature_cols = [
    'Store','Dept','Size','Temperature','Fuel_Price','CPI','Unemployment','Total_MarkDown',
    'Month','Year','WeekOfYear','Quarter','IsHoliday',
    'Weekly_Sales_Lag1','Weekly_Sales_Lag2','Rolling_Mean_4','Rolling_Std_4'
] + [c for c in tr.columns if c.startswith('Store_Type_')]

X_train, y_train = tr[feature_cols].copy(), tr['Weekly_Sales']
X_val, y_val = val[feature_cols].copy(), val['Weekly_Sales']

for c in ['IsHoliday']:
    X_train[c] = X_train[c].astype(int)
    X_val[c] = X_val[c].astype(int)

xgb_model = xgb.XGBRegressor(
    n_estimators=600, max_depth=8, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8, random_state=42, n_jobs=-1
)
xgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

xgb_preds = xgb_model.predict(X_val)
xgb_preds = np.clip(xgb_preds, 0, None)  # sales can't be negative
xgb_wmae = wmae(y_val, xgb_preds, val['IsHoliday'])
print(f"XGBoost WMAE: {xgb_wmae:,.2f}")
```

```python
importances = pd.Series(xgb_model.feature_importances_, index=feature_cols).sort_values()
plt.figure(figsize=(8,7))
importances.plot(kind='barh')
plt.title('XGBoost Feature Importance')
plt.xlabel('Importance')
plt.tight_layout(); plt.show()
```

### Ablation Study
Measuring how much of XGBoost's accuracy comes from lag/rolling features specifically.

```python
# --- Ablation: how much does XGBoost actually rely on lag/rolling features? ---
no_lag_cols = [c for c in feature_cols if c not in
               ['Weekly_Sales_Lag1','Weekly_Sales_Lag2','Rolling_Mean_4','Rolling_Std_4']]

X_train_nolag, X_val_nolag = tr[no_lag_cols].copy(), val[no_lag_cols].copy()
X_train_nolag['IsHoliday'] = X_train_nolag['IsHoliday'].astype(int)
X_val_nolag['IsHoliday'] = X_val_nolag['IsHoliday'].astype(int)

xgb_nolag = xgb.XGBRegressor(
    n_estimators=600, max_depth=8, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8, random_state=42, n_jobs=-1
)
xgb_nolag.fit(X_train_nolag, y_train, eval_set=[(X_val_nolag, y_val)], verbose=False)

preds_nolag = np.clip(xgb_nolag.predict(X_val_nolag), 0, None)
wmae_nolag = wmae(y_val, preds_nolag, val['IsHoliday'])

print(f"XGBoost WITH lag/rolling features:    WMAE = {xgb_wmae:,.2f}")
print(f"XGBoost WITHOUT lag/rolling features: WMAE = {wmae_nolag:,.2f}")
print(f"Lag/rolling features account for {(wmae_nolag - xgb_wmae):,.2f} WMAE "
      f"({(wmae_nolag - xgb_wmae) / wmae_nolag:.1%} relative improvement)")
```

### Rolling-Origin Cross-Validation
Re-validating across 3 time windows to confirm results aren't a lucky split.

```python
# --- Rolling-origin cross-validation ---
def make_split(full_df, n_val_weeks, offset_weeks):
    max_date = full_df['Date'].max()
    val_end = max_date - pd.Timedelta(weeks=offset_weeks)
    val_start = val_end - pd.Timedelta(weeks=n_val_weeks)
    tr_split = full_df[full_df['Date'] <= val_start].copy()
    val_split = full_df[(full_df['Date'] > val_start) & (full_df['Date'] <= val_end)].copy()
    return tr_split, val_split

cv_offsets = [0, 12, 24]
cv_results = {'Seasonal Naive': [], 'XGBoost': []}

for offset in cv_offsets:
    tr_cv, val_cv = make_split(train, n_val_weeks=12, offset_weeks=offset)
    if len(val_cv) == 0 or len(tr_cv) == 0:
        continue

    naive_preds_cv = seasonal_naive_predict(tr_cv, val_cv)
    cv_results['Seasonal Naive'].append(wmae(val_cv['Weekly_Sales'], naive_preds_cv, val_cv['IsHoliday']))

    X_tr_cv = tr_cv[feature_cols].copy(); X_tr_cv['IsHoliday'] = X_tr_cv['IsHoliday'].astype(int)
    X_val_cv = val_cv[feature_cols].copy(); X_val_cv['IsHoliday'] = X_val_cv['IsHoliday'].astype(int)

    model_cv = xgb.XGBRegressor(
        n_estimators=600, max_depth=8, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8, random_state=42, n_jobs=-1
    )
    model_cv.fit(X_tr_cv, tr_cv['Weekly_Sales'], verbose=False)
    preds_cv = np.clip(model_cv.predict(X_val_cv), 0, None)
    cv_results['XGBoost'].append(wmae(val_cv['Weekly_Sales'], preds_cv, val_cv['IsHoliday']))

    print(f"Fold offset={offset}w | val window {val_cv['Date'].min().date()} to {val_cv['Date'].max().date()} "
          f"| Naive WMAE={cv_results['Seasonal Naive'][-1]:,.0f} | XGBoost WMAE={cv_results['XGBoost'][-1]:,.0f}")

print()
for model_name, scores in cv_results.items():
    scores = np.array(scores)
    print(f"{model_name}: mean WMAE = {scores.mean():,.2f}  (std = {scores.std():,.2f}, folds = {len(scores)})")
```
